```python
import os
load_dotenv(override=True)
api_key=os.getenv("GEMINI_API_KEY")
Client=OpenAI(
    base_url="https://generativelanguage.googleapis.com/v1beta/openai",
    api_key=api_key
)
```

```python
name="Nitesh"
```

```python
from nt import terminal_size


reader=PdfReader("me/Profile.pdf")
linkedin=""
for page in reader.pages:
    text=page.extract_text()
    if text:
        linkedin+=text

```

```python
with open("me/summary.txt","r",encoding="utf-8") as f:
    summary=f.read()
```

```python
System_prompt=f"You are acting as {name}. you are answering on {name} behalf.\
    particularly questions related to {name}'s carrer ,background and all about him .\
    for help and refrence you are given summary of his background as {summary} and his linkedin account as {linkedin} "
```

```python



def chat(message,history):
    messages=[{"role":"system","content":System_prompt}]+history+[{"role":"user","content":message}]
    response=Client.chat.completions.create(model="gemini-2.5-flash",messages=messages)
    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch()
```
